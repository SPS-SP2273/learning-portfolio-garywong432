```python
def greeting(name):
    if name == 'Batman':
        return 'Hello Batman! So, nice to meet you!'
    return f'Hello {name}!'
```

The function works without a else statement because the "return f'Hello {name}" is essentially the else statement. The code works when greeting(), and a name is put into the brackets, if the name is Batman, then it will return "Hello Batman! So, nice to meet you!", if not then it will give back "Hello (with whatever name that was given)"

```python
def calculate_bmi(weight, height):
    if weight <= 0:
        return 'Please give positive integer tqvmuch'
    elif height <= 0:
        return 'Please give positive integer tqvmuch'
    BMI = weight / (height**2)
    if BMI < 18.5:
        return 'Underweight'
    elif 18.5 <= BMI < 25:
        return 'Normal Weight'
    elif 25 <= BMI <=29.9:
        return 'Overweight'
    elif BMI > 29.9:
        return 'OBESE'
    return 'Please input proper values'


calculate_bmi(-50,1.416766)
```

```python
def factorial(n):
    try:
        number = int(n)
        if number < 0:
            return "Please give a positive integer"
        elif n == 0:
            return 1
        else:
            result = 1
            for number2 in range(1, number+1):
                result *= number2
            return result
    
    except:TypeError
    return "Please use an integer"



print(factorial(7))
print(factorial('wewe'))
```
