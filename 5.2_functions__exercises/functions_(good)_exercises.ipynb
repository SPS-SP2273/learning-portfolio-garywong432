```python
def convert_celsius(temp_celsius, target_scale = "Fahrenheit"):
    kelvin =temp_celsius + 273.15
    fahrenheit = 1.8 * temp_celsius + 32
    if target_scale == "Kelvin":
        print(kelvin, "Kelvin")
    else:
        print(fahrenheit, "Fahrenheit") 


convert_celsius(100,"Kelvin")
```

```python
def convert_fahrenheit(temp_fahrenheit, target_scale = "Celsius"):
    celsius = (temp_fahrenheit - 32)*5/9
    kelvin = celsius + 273.15
    if target_scale == "Kelvin":
        print(kelvin, "K")
    else:
        print(celsius, "C")

convert_fahrenheit(100, "Kelvin")
```

```python
def convert_temperature():
    temp = int(input("What is your temperature"))
    source_scale = input("Celsius, Fahrenheit or Kelvin")
    target_scale = input("Convert to Celsius, Fahrenheit or Kelvin?")
    if source_scale == "Celsius":
        if target_scale == "Kelvin":
            print(temp+273.15, "K")
        else:
            print(1.8*temp + 32, "F")
    elif source_scale == "Fahrenheit":
        if target_scale == "Celsius":
            print((temp-32)*5/9, "C")
        else:
            print((temp-32)*5/9 + 273.15, "K")
    elif source_scale == "Kelvin":
        if target_scale == "Celsius":
            print(temp - 273.15, "C")
        else:
            print((temp-273.15)*1.8 +32, "F")
    else:
        print("Invalid input")

convert_temperature()
```
