```python
%%timeit
for number in range(1, 10):
    if number % 2 == 0:
        continue

    square = number ** 2
    print(square)
```

```python
my_list = []
for number in range(1, 10):
    if number % 2 == 0:
        continue

    square = number ** 2
    my_list.append(square)
    print(square)
print(my_list)
```

```python
%%timeit
[number**2 for number in range(1, 10) if number % 2 != 0]
```

```python
co2_data = [
    (2000, 369.55), (2001, 371.14), (2002, 373.28), 
    (2003, 375.80), (2004, 377.52), (2005, 379.80), 
    (2006, 381.90), (2007, 383.79), (2008, 385.60), 
    (2009, 387.43), (2010, 389.90), (2011, 391.65), 
    (2012, 393.85), (2013, 396.52), (2014, 398.65),
    (2015, 400.83), (2016, 404.24), (2017, 406.55), 
    (2018, 408.52), (2019, 411.44), (2020, 414.24)
]

for year, co2 in co2_data:
    if co2_data.index((year, co2)) == 0: # start at first entry
        previous_co2 = co2 # set previous co2 to current co2 
    change = co2 - previous_co2
    #print(change) was checking if the code was working
    if change > 3:
        print(f'Year: {year}, Change: {change}ppm')
    previous_co2 = co2 # redefine "previous_co2" such that "co2" value can be set to the next entry
    
```
