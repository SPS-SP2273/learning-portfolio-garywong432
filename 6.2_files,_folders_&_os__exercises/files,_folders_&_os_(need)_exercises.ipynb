```python
dictionary = {
'jan': '01', 'feb': '02', 'mar': '03',
'apr': '04', 'may': '05', 'jun': '06',
'jul': '07', 'aug': '08', 'sep': '09',
'oct': '10', 'nov': '11', 'dec': '12'
}

filename = 'oct-08_singapore.txt'
month = filename.split('-')[0]
remainder = filename.split('-')[1]
new_month = dictionary[month]

remainder2 = remainder.split('_')[0]
remainder3 = remainder.split('_')[1]

print(f'{new_month}-{remainder2}-{remainder3}')
```

```python
def rename_my_file(old_file_name):
    filename = old_file_name
    month = filename.rsplit('-',1)[0]
    remainder = filename.split('-')[1]
    new_month = dictionary[month]

    remainder2 = remainder.split('_')[0]
    remainder3 = remainder.split('_')[1]

    return f'{new_month}-{remainder2}-{remainder3}'


rename_my_file('feb-08_singapore.txt')
```

```python
dict_1 = {'jan': '01', 'feb': '02', 'mar': '03',
'apr': '04', 'may': '05', 'jun': '06',
'jul': '07', 'aug': '08', 'sep': '09',
'oct': '10', 'nov': '11', 'dec': '12'
}

def rename_my_file(old_file_name):
    # the function will run the code if the file format is my desired one, if not it will return the old file name
    try:
        filename = old_file_name
        month = filename.split('-')[0]
        remainder = filename.split('-')[1]
        new_month = dict_1[month]

        remainder2 = remainder.split('_')[0]
        remainder3 = remainder.split('_')[1]

        new_file2 = f'{new_month}-{remainder2}-{remainder3}'
        return new_file2
    except:
        return old_file_name

all_data_list = glob.glob('testest/os-collaboration-exercise-data/**/*.txt', recursive= True)
print(all_data_list)

for i in all_data_list:
    separated_file_name = i.split('\\')[-1]
    new_file_names = rename_my_file(separated_file_name)
    destination = os.path.join('testest/os-collaboration-exercise-data', new_file_names)
    os.rename(i, destination)


```
