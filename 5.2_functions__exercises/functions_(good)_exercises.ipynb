```python
def convert_fahrenheit(temp_fahrenheit, target_scale = "Celsius"):
    celsius = (temp_fahrenheit - 32)*5/9
    kelvin = celsius + 273.15
    if target_scale == "Kelvin":
        return f'{kelvin}K'
    else:
        return f'{celsius}C'

convert_fahrenheit(100, "Kelvin")
```

```python
def convert_kelvin(temp_kelvin, target_scale = 'Celsius'):
    celsius = temp_kelvin - 273.15
    fahrenheit = (celsius * 9/5) +32
    if target_scale == 'Celsius':
        return f'{celsius} C'
    else:
        return f'{fahrenheit}F' 


def convert_temp(temp, current_units, target_units):
    if current_units == 'Celsius':
        return convert_celsius(temp, target_units)
    elif current_units == 'Kelvin':
        return convert_kelvin(temp,target_units)
    elif current_units == 'convert_fahrenheit':
        return convert_fahrenheit(temp, target_units)



convert_temp(100,'Kelvin', 'Celsius')
```
